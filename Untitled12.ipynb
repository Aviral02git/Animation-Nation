````python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error,r2_score
import numpy as np
import matplotlib.pyplot as plt
````

````python
gen = pd.read_csv("Plant_1_Generation_Data.csv")
weather = pd.read_csv("Plant_1_Weather_Sensor_Data.csv")
````

````python
print(gen.shape)
print(weather.shape)
gen.head()
````

````python
gen["DATE_TIME"] = pd.to_datetime(gen["DATE_TIME"])
weather["DATE_TIME"] = pd.to_datetime(weather["DATE_TIME"])
````

````python
df = pd.merge(gen, weather, on=["DATE_TIME", "PLANT_ID"])
df.head()
````

````python
df.isnull().sum()
````

````python
df["hour"] = df["DATE_TIME"].dt.hour
df["day"] = df["DATE_TIME"].dt.day
df["month"] = df["DATE_TIME"].dt.month
````

##Target (what we predict)
DC_POWER

```

```



````python
features = ["IRRADIATION", "AMBIENT_TEMPERATURE", "MODULE_TEMPERATURE", "hour"]
X = df[features]
y = df["DC_POWER"]
````

````python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)
````

````python
model = RandomForestRegressor(n_estimators=100, random_state=42)
model.fit(X_train, y_train)
````

````python
preds = model.predict(X_test)

mae = mean_absolute_error(y_test, preds)
rmse = np.sqrt(mean_squared_error(y_test, preds))
r2 = r2_score(y_test, preds)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2 Score:", r2)
````

````python
plt.figure(figsize=(12,5))
plt.plot(y_test.values[:200], label="Actual")
plt.plot(preds[:200], label="Predicted")
plt.legend()
plt.title("Solar Power Prediction")
plt.show()
````

````python
import joblib
joblib.dump(model, "solar_model.pkl")
````
